```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.datasets import load_iris
```

```python

num =np.linspace(0,60,100)
fig =plt.figure()
ax =plt.axes() 
plt.subplot(2,1,1)
plt.plot(num,np.sin(num),color='r',linestyle=':')
plt.subplot(2,1,2)
plt.plot(num,np.cos(num))

```

```python
plt.subplot(2,2,1)
plt.plot(num,np.sin(num),color='k',linestyle=':')
plt.subplot(2,2,2)
plt.plot(num,np.cos(num),color='g',linestyle='--')
plt.subplot(2,2,3)
plt.plot(num,np.tan(num),color='r')
plt.subplot(2,2,4)
plt.plot(num,np.sinh(num))
```

```python
plt.axis([-1,15,-1.5,1.5])
plt.plot(num,np.sin(num),linestyle='--')
plt.plot(num,np.cos(num))
plt.title('the_chart',fontsize=16,color='r')
plt.xlabel('x_label')
plt.ylabel('y_label')
plt.legend(['sin(x)','cos(x)'])
```

```python
import pandas as pd
df=pd.DataFrame({'month':['san','mon','tue','wed','thu'],
                 'salary':[1200,2000,400,700,5000]})
df
```

```python
fig, ax = plt.subplots()
fruits = ['apple', 'blueberry', 'banana', 'orange']
count = [40, 100, 30, 55]
bar_labels = ['red', 'blue', 'yellow', 'orange']
ax.bar(fruits,
       count,
       label='bar_labels',
       color = ['red', 'blue', 'yellow', 'orange'])
ax.set_ylabel('fruit supply')
ax.set_xlabel('fruits names')
ax.set_title('Fruit supply by kind and color',fontsize=16,color="k")
ax.legend(shadow=True)

```

```python
fig , ax = plt.subplots()
fruits = ['apple', 'blueberry', 'banana', 'orange']
counts = [40, 100, 30, 55]
bar_labels = ['red', 'blue', 'yellow', 'orange']

ax.barh(fruits, counts,label = bar_labels, color = ['red', 'blue', 'yellow', 'orange'] )
ax.set_ylabel('fruit supply')
ax.set_xlabel('fruits names')

ax.set_title('Fruit supply by kind and color')
ax.legend(shadow = True)
```
